```python
import numpy as np
import pandas as pd
```

```python
# Set random seed for reproducibility
np.random.seed(42)

# Create a raw, messy dataset
data = {
    "Employee_ID": [
        "EMP01",
        "EMP02",
        "EMP03",
        "EMP04",
        "EMP01",
        "EMP05",
        "EMP06",
        "EMP07",
        "EMP08",
        "EMP09",
    ],
    "Age": [28, np.nan, 45, 33, 28, 52, "thirty-one", 41, 19, 62],
    "Join_Date": [
        "2023-01-15",
        "2022-11-05",
        "12/04/2021",
        "2024-02-18",
        "2023-01-15",
        np.nan,
        "2020-08-22",
        "2025-01-01",
        "2023-06-30",
        "invalid_date",
    ],
    "Salary": [
        55000,
        62000,
        85000,
        -4000,
        55000,
        110000,
        72000,
        np.nan,
        48000,
        95000,
    ],
    "Department": [
        "HR",
        "IT",
        "Fin",
        "Marketing",
        "HR",
        "IT",
        "Sales",
        "  HR  ",
        "Fin",
        "IT",
    ],
}
```

```python
data
```

to create a dataframe from data (dict)

```python
df = pd.DataFrame(data)
```

```python
df.info()
```

```python
df.head()
```

```python
#to insert new data or  row

# Create the new row as a mini DataFrame
new_row = pd.DataFrame([[np.nan, 0, "?", 65000, "IT"]], columns=df.columns)

# Insert at index position 0 (the very top)
df = pd.concat([new_row, df]).reset_index(drop=True)
```

```python
df.tail()
```

```python
df.head()
```

```python
# Insert at the very end using the length of the dataframe
df.loc[len(df)] = ["EMP10", 35, "2026-03-01", "?", "IT"]
```

```python
df
```

```python
df.info()
```

```python
from numpy._core.defchararray import count
df.isnull().sum()
```

```python
df.replace(to_replace='?', value =np.nan,  inplace =True)
```

```python
df
```

```python
#replace null values in salary column with mean of that column
df['Salary'].replace(to_replace =np.nan , value = np.mean(df['Salary']) ,  inplace =True)
```

```python
df
```

```python
df['Join_Date'].replace(to_replace=[np.nan , "invalid_date"] , value= "0-0-0" , inplace = True)
```

```python
df
```

```python
df['Age'].replace(to_replace="thirty-one" , value= 0, inplace = True)
```

```python
df['Age'].replace(to_replace=np.nan , value= np.mean(df['Age']) , inplace = True)
```

```python
df
```

```python
#fill null values with custom values
df.fillna(value = {'Age' : df['Age'].mean() , 'Salary' : df['Salary'].median()})
```

## Dealing With Catagorical Data

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
data_url = 'http://vincentarelbundock.github.io/Rdatasets/csv/carData/Salaries.csv'
#with using index_col =0 will drop extra index column
df = pd.read_csv(data_url, index_col =0 )
df
```

```python
df_dummies = pd.get_dummies(df[['rank', 'sex']])
df_dummies
```

```python
df_dummies = pd.get_dummies(df[['rank', 'sex']] , drop_first = True)
df_dummies
```

```python
df2_dummies = pd.get_dummies(df[['rank', 'sex' , 'discipline']] , drop_first = True )
df2_dummies
```

```python
df3 = pd.concat([df2_dummies , df] , axis=1)
df3
```

```python
df3.drop(['rank' , 'sex', 'discipline'] , axis =1 , inplace =True)
```

```python
df3
```

```python
x =  df3.drop('salary' , axis =1).values
y = df3['salary']
```

```python
x
```

```python
y
```

```python
x_train , x_test , y_train , y_test = train_test_split(x ,y, test_size=0.2 , random_state =42)
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```
